```python
#import libraries
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

transform = transforms.ToTensor()

train_data = datasets.FashionMNIST(root="data", train=True, download=True, transform=transform)
test_data  = datasets.FashionMNIST(root="data", train=False, download=True, transform=transform)

train_batch = DataLoader(
    train_data,
    shuffle = True,
    batch_size = 256
)

test_batch = DataLoader(
    test_data,
    shuffle = False,
    batch_size = 256
)

class Fashionista(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(784,392)
        self.layer2 = nn.Linear(392,196)
        self.layer3 = nn.Linear(196,98)
        self.layer4 = nn.Linear(98,49)
        self.layer5 = nn.Linear(49,10)
        self.ReLU = nn.ReLU()

    def forward(self,x):
        x = self.ReLU(self.layer1(x))
        x = self.ReLU(self.layer2(x))
        x = self.ReLU(self.layer3(x))
        x = self.ReLU(self.layer4(x))
        x = self.layer5(x)

        return x
    
model = Fashionista()
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)


model.train()

for epoch in range(20):
    total_loss = 0

    for images,labels in train_batch:
        images = images.reshape(-1,784)

        predictions = model(images)

        loss = criterion(predictions,labels)

        optimizer.zero_grad()

        loss.backward()
        
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_batch)
    print(f"Epoch {epoch + 1}. Loss per batch: {average_loss}")

model.eval()
correct = 0
total_images = 0
with torch.no_grad():
    for images, labels in test_batch:
        images = images.reshape(-1,784)
        predictions = model(images)
        machine_prediction = torch.argmax(predictions, dim=1)
        total_images += labels.size(0)
        correct += (machine_prediction == labels).sum()

accuracy = 100 * (correct/total_images)

torch.save(model.state_dict(), "fashion_model.pth")
print("Updated model saved!")
print(f"Accuracy: {accuracy}%")


```
